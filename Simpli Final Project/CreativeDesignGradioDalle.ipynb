```python
import gradio as gr
from openai import OpenAI
import requests
from io import BytesIO
from gradio_client import Client
# removed

```

```python

def generate_image(prompt):
    """
    Generate an image based on a text prompt using OpenAI's DALL-E.
    """
    client = OpenAI(api_key = openai.api_key)

    try:
        # Generate image using OpenAI's DALL-E API
        response = client.images.generate(
            model="dall-e-3",
            prompt=prompt
        )
        # Parse response to get image URL
        image_url = response.data[0].url  # Assuming URL is in response content
        
        # Fetch the image from the URL
        image_response = requests.get(image_url)
        image = Image.open(BytesIO(image_response.content))
        
        return image
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
# Set up the Gradio interface
interface = gr.Interface(
    fn=generate_image,
    inputs=gr.Textbox(label="Enter your text prompt"),
    outputs=gr.Image(type="pil", label="Generated Image"),
    title="Bespoke Image Generator",
    description="Enter a text prompt to generate a custom banner or poster design using AI."
)

# Launch the interface
interface.launch()
```
